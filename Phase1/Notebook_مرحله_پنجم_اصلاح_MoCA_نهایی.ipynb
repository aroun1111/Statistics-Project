## بررسی توزیع متغیرهای مهم برای شناخت نمونه

در این قسمت به‌جای رسم نمودار برای تمام ستون‌ها، چند متغیر را انتخاب می‌کنیم که ابتدا به شناخت نمونه و وضعیت شناختی افراد کمک می‌کنند. انتخاب نمودارها مرحله‌به‌مرحله انجام شده است: ابتدا سن، سپس نمره‌های شناختی و بعد متغیرهای مرتبط با سؤال احتمالی پژوهش.

برای نمایش و آمار توصیفی MoCA، فعلاً فقط مقادیر بین 0 تا 30 استفاده می‌شوند. کدهای 88 و 4- و همچنین مقادیر بالاتر از 30 در این بررسی وارد نمی‌شوند. داده اصلی تغییر نمی‌کند و تعداد مقادیر خارج از این محدوده جداگانه گزارش شده است. این فیلتر موقت است و باید با راهنمای رسمی متغیرها تأیید شود.

```python
Data[['NACCAGE' , 'CDRSUM' , 'CDRGLOB']].describe().T

MOCATOTS_valid = Data[Data['MOCATOTS'].between(0, 30)]
MOCATOTS_valid['MOCATOTS'].describe()
```

```python
MOCATOTS_valid = Data[Data['MOCATOTS'].between(0, 30)]

fig, axes = plt.subplots(2, 2, figsize=(12, 8))

sns.histplot(data=Data, x='NACCAGE', bins=20, kde=True, ax=axes[0, 0], color='lightblue')
axes[0, 0].set_title('Age distribution')
axes[0, 0].set_xlabel('NACCAGE')

sns.histplot(data=MOCATOTS_valid, x='MOCATOTS', bins=20, kde=True, ax=axes[0, 1], color='lightblue')
axes[0, 1].set_title('MoCA score distribution (0-30)')
axes[0, 1].set_xlabel('MOCATOTS')

sns.histplot(data=Data, x='CDRSUM', bins=20, ax=axes[1, 0], color='lightblue')
axes[1, 0].set_title('CDR Sum of Boxes distribution')
axes[1, 0].set_xlabel('CDRSUM')

sns.countplot(data=Data, x='CDRGLOB', order=sorted(Data['CDRGLOB'].dropna().unique()), ax=axes[1, 1], color='lightblue')
axes[1, 1].set_title('Global CDR categories')
axes[1, 1].set_xlabel('CDRGLOB')

plt.tight_layout()
plt.show()
```

## بررسی متغیرهای منتخب MRI

فایل MRI تعداد زیادی متغیر مربوط به نواحی مختلف مغز دارد. رسم نمودار برای همه آن‌ها خوانایی تحلیل را کم می‌کند. بنابراین در این مرحله فقط ضخامت متوسط قشر انتورینال در نیمکره چپ و راست را بررسی می‌کنیم؛ چون این متغیرها به موضوع احتمالی پژوهش ارتباط مستقیم‌تری دارند.

این انتخاب به معنی کنار گذاشتن دائمی سایر نواحی مغز نیست؛ در صورت نیاز و با توجیه پژوهشی می‌توان در مراحل بعدی متغیرهای دیگری را بررسی کرد.

```python
Data[['LH_ENTORHINAL_AVGTH' , 'RH_ENTORHINAL_AVGTH']].describe().T
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

sns.histplot(data=Data, x='LH_ENTORHINAL_AVGTH', bins=25, kde=True, ax=axes[0], color='lightblue')
axes[0].set_title('Left entorhinal thickness')
axes[0].set_xlabel('LH_ENTORHINAL_AVGTH')

sns.histplot(data=Data, x='RH_ENTORHINAL_AVGTH', bins=25, kde=True, ax=axes[1], color='lightblue')
axes[1].set_title('Right entorhinal thickness')
axes[1].set_xlabel('RH_ENTORHINAL_AVGTH')

plt.tight_layout()
plt.show()
```

```python
ENTORHINAL = Data[['LH_ENTORHINAL_AVGTH' , 'RH_ENTORHINAL_AVGTH']].melt(var_name='Hemisphere', value_name='Thickness')

plt.figure(figsize=(8, 5))
sns.boxplot(data=ENTORHINAL, x='Hemisphere', y='Thickness', color='lightblue')
plt.title('Entorhinal thickness by hemisphere')
plt.xlabel('Hemisphere')
plt.ylabel('Average thickness')
plt.tight_layout()
plt.show()
```

## بررسی ارتباط اولیه ضخامت انتورینال با وضعیت شناختی

پس از شناخت توزیع نمره‌های شناختی و ضخامت انتورینال، می‌توانیم بررسی کنیم که آیا بین این متغیرها الگوی اولیه‌ای دیده می‌شود یا نه. چون CDRGLOB چند سطح مشخص دارد، مقایسه توزیع ضخامت انتورینال در این گروه‌ها از یک نمودار پراکندگی ساده خواناتر است.

این نمودار توصیفی است و به‌تنهایی رابطه علّی یا اثر میانجی‌گری را نشان نمی‌دهد.

```python
Data['ENTORHINAL_AVGTH'] = Data[['LH_ENTORHINAL_AVGTH' , 'RH_ENTORHINAL_AVGTH']].mean(axis=1)

plt.figure(figsize=(9, 5))
sns.boxplot(data=Data, x='CDRGLOB', y='ENTORHINAL_AVGTH', color='lightblue')
sns.stripplot(data=Data, x='CDRGLOB', y='ENTORHINAL_AVGTH', color='black', alpha=0.18, size=2)
plt.title('Entorhinal thickness across global CDR categories')
plt.xlabel('CDRGLOB')
plt.ylabel('Mean entorhinal thickness')
plt.tight_layout()
plt.show()
```

```python
sns.scatterplot(data=Data, x='NACCAGE', y='CDRSUM', alpha=0.35, color='steelblue')
plt.title('Age and CDR Sum of Boxes')
plt.xlabel('NACCAGE')
plt.ylabel('CDRSUM')
plt.tight_layout()
plt.show()
```
